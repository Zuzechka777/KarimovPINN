```python
import numpy as np
import matplotlib.pyplot as plt
import time
import torch
import torch.nn as nn
import tqdm
from scipy.interpolate import griddata
```

```python
data = np.loadtxt(
    "PINN2026-1/tasks/1_copy_func_model/data/muT_turbulent_plain_jet.txt",
    comments="%"
)
```

```python
x = data[:, 0]
y = data[:, 1]
u = data[:, 2]
print(data.shape)
print(x.min(), x.max())
print(y.min(), y.max())
print(u.min(), u.max())
```

```python
x_grid = np.linspace(x.min(), x.max(), 100)
y_grid = np.linspace(y.min(), y.max(), 100)
x_grid, y_grid = np.meshgrid(x_grid, y_grid)
u_grid = griddata((x,y),u, (x_grid, y_grid), method = "nearest")   
```

```python
plt.pcolor(x_grid, y_grid, u_grid, shading="auto")
plt.colorbar()
```

```python
x_norm = x_grid/x_grid.max()
y_norm = y_grid/y_grid.max()
u_norm = u_grid/u_grid.max()
```
